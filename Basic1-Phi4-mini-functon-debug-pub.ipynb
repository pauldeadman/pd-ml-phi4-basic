```python
import torch
import json
import random
import string
import re
from transformers import AutoModelForCausalLM, AutoProcessor, GenerationConfig,pipeline,AutoTokenizer

model_path = "microsoft/Phi-4-mini-instruct"

model = AutoModelForCausalLM.from_pretrained(
    model_path,
    device_map="cuda",
    attn_implementation="eager",
    torch_dtype="auto",
    trust_remote_code=True)

tokenizer = AutoTokenizer.from_pretrained(model_path, trust_remote_code=True)
```

```python
# Tools should be a list of functions stored in json format
tools = [
    {
        "name": "get_match_result",
        "description": "get match result",
        "parameters": {
            "match": {
                "description": "The name of the match",
                "type": "str",
                "default": "Arsenal vs ManCity"
            }
        }
    },
]

# Function implementations

def get_match_result(match: str) -> str:
    # This would be replaced by a weather API
    match_data = {
        "Arsenal vs ManCity": "1:1",
        "Chelsea vs ManUnited": "0:2"
    }
    return match_data.get(match, "I don't know")


messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant",
        "tools": json.dumps(tools), # pass the tools into system message using tools argument
    },
    {
        "role": "user",
        "content": "What is the result of Arsenal vs ManCity today?"
    }
]
```

```python
#========
# https://huggingface.co/docs/transformers/en/chat_templating#using-applychattemplate
# Converts list of role:, content: ... prompts into native chat prompts for ROLES: user [from user], assistant [from model], system [how model should act - prefix], tool_call [requests for help]
inputs = tokenizer.apply_chat_template(messages, add_generation_prompt=True, return_dict=True, return_tensors="pt")
#========
print("tokenizer.apply_chat_template:")
print(inputs)
#print(tokenizer.decode(inputs["input_ids"][0])
#========
# iterate over dictionary key,value pairs [for k, v in inputs.items()]
# adds the CPU/GPU parameter to each "device='cuda:0'"
inputs = {k: v.to(model.device) for k, v in inputs.items()}
print("model.device")
print(inputs)
#========
# Run LLM inference
# Pass in inputs as list of parameters input_ids, attention_mask, followed by max_new_tokens
output = model.generate(**inputs, max_new_tokens=128)
print("model.generate [output]")
print(output)
#========
# return prompt from model (https://huggingface.co/docs/transformers/en/chat_templating#using-applychattemplate)
print("tokenizer.decode")
# Get length of inputs_ids tensor
len1 = len(inputs["input_ids"][0])
print("Len1=",len1)
#p
rint(tokenizer.decode(output[0][len(inputs["input_ids"][0]):]))
# Get output, which includes input at start
# Decode output[0] which is an tensor/array starting from len1 to end
print(tokenizer.decode(output[0][len1:]))

```

```python
tokenizer.batch_decode(output)

#========
# get response in output[0] tensor from length of input
response = tokenizer.decode(output[0][len(inputs["input_ids"][0]):], skip_special_tokens=True)
print("resoonse")
print(response)

#========
# Generate a unique key
tool_call_id = ''.join(random.choices(string.ascii_letters + string.digits, k=9))

#========
# Create a tool_calls prompt and append to list of prompts
messages.append({"role": "assistant", "tool_calls": [{"type": "function", "id": tool_call_id, "function": response}]})

#========
# Convert response into tool_call
try :
    tool_call = json.loads(response)[0]

except :
    json_part = re.search(r'\[.*\]', response, re.DOTALL).group(0)

    tool_call = json.loads(json_part)[0]


#========
# Get the tool name to call
function_name = tool_call["name"]
print("function_name:",function_name)

#========
# Get the list of arguments and pass to tool
# note that should check function name!
arguments = tool_call["arguments"]

result = get_match_result(**arguments)

#========
# Create a tool (aka reponse) prompt and append to list of prompts
messages.append({"role": "tool", "tool_call_id": tool_call_id, "name": "get_match_result", "content": str(result)})

#========
# print full list of prompts
print(messages)
```
